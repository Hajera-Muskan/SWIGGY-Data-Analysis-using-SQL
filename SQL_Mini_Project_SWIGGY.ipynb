```python
%sql
SELECT 
    n.campaign_id,
    SUM(CASE WHEN m.ad_clicked_flag THEN 1 ELSE 0 END) AS total_clicks,
    SUM(CASE WHEN m.campaign_exposed_flag THEN 1 ELSE 0 END) AS total_impressions,
    (SUM(CASE WHEN m.ad_clicked_flag THEN 1 ELSE 0 END) * 100.0 / NULLIF(SUM(CASE WHEN m.campaign_exposed_flag THEN 1 ELSE 0 END), 0)) AS ctr_percentage
FROM workspace.swiggy.swiggy_fact_transactions m
JOIN workspace.swiggy.swiggy_dim_campaign n
    ON m.campaign_id = n.campaign_id
GROUP BY n.campaign_id
ORDER BY ctr_percentage DESC;
```

```python
import pandas as pd

fact = spark.table("workspace.swiggy.swiggy_fact_transactions_v_2").toPandas()
fact["transaction_date"] = pd.to_datetime(fact["transaction_date"])
dim_restaurant_product = spark.table("workspace.swiggy.swiggy_dim_restaurant_product").toPandas()
today = fact["transaction_date"].max()

c3 = fact.merge(
     dim_restaurant_product[["restaurant_product_id", "cuisine_tag"]],
     how = "inner",
     on= "restaurant_product_id"
 )
c3["year_month"]=c3["transaction_date"].dt.strftime("%Y-%m")

c3_base =(
     c3.groupby(["city", "year_month","cuisine_tag"], as_index=False).agg(net_revenue=("net_amount","sum"), orders=("transaction_id","count"))
 )

c3_base = c3_base.sort_values(["city","year_month","net_revenue"], ascending=[True,True,False])
c3_base["rn"]= c3_base.groupby(["city","year_month"]).cumcount()+1
c3_out = c3_base[c3_base["rn"]<=5].copy()
c3_out["net_revenue"]= c3_out["net_revenue"].round(2)


display(c3_out.sort_values(["year_month","city","rn"]).head(200))
```

```python
cutoff_60 = today - pd.Timedelta(days=60)

c4 = fact[fact["transaction_date"]>= cutoff_60].merge(
    dim_restaurant_product[["restaurant_product_id","restaurant_name"]],
    how="inner",
    on = "restaurant_product_id"
    )

c4_out =(
    c4.groupby(["city","restaurant_name"], as_index= False).agg(
        orders=("transaction_id", "count"),
        net_revenue=("net_amount","sum"),
        avg_delivery_minutes=("delivery_minutes","mean")
    )
)

c4_out["net_revenue"]= c4_out["net_revenue"].round(2)
c4_out["avg_delivery_minutes"]= c4_out["avg_delivery_minutes"].round(2)

display(c4_out.sort_values(["city", "net_revenue"], ascending=[True, False]).head(200))
```
